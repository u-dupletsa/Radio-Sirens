```python
ska_color_pink = '#d60068'
ska_color_intermediate = '#840068'
ska_color_purple = '#180068'
radio_color = '#f5e15f'
```

```python
default_kwargs = dict(
    bins=100,
    smooth=2,
    smooth1d=2,

    title_fmt='.1f',
    title_kwargs=dict(fontsize=18, pad=2),
    label_kwargs=dict(color='black', fontsize=18),

    levels=(1 - np.exp(-0.5), 1 - np.exp(-2)),
    plot_density=False,
    plot_datapoints=False,
    max_n_ticks=3,
    labelpad=-0.02
)
```

```python
cosmo_corner_lbs = [r'$H_0\,[\mathrm{km}\,\mathrm{s}^{-1}\,\mathrm{Mpc}^{-1}]$',
              r'$\Omega_{m,0}$', 
              r'$\gamma$', 
              r'$\kappa$', 
              r'$z_p$',
              r'$b_{\rm GW}$',
              r'$\alpha_{\rm GW}$']
cosmo_title_labels = [r'$H_0$',
              r'$\Omega_{m,0}$', 
              r'$\gamma$', 
              r'$\kappa$', 
              r'$z_p$',
              r'$b_{\rm GW}$',
              r'$\alpha_{\rm GW}$']
```

```python
def plot_corner(ref_data, data_hi_star, label, injected_values, labels=cosmo_corner_lbs, title_labels=cosmo_title_labels, lbl_size=16):
    figure = corner.corner(ref_data, 
                           titles = title_labels, labels = labels,
                           color=ska_color_purple, alpha=1,
                           range=[(40, 100), (0, 1), (1, 5), (2.5, 9), (0.75, 1.25), (0, 3.5), (-1, 1)],  # one tuple per parameter
                           rect=[0, 0, 1, 1], 
                           fill_contours=False,
                           show_titles=False,
                           truths=injected_values,
                           truth_color=ska_color_pink,
                           title_quantiles=[0.16, 0.5, 0.84],
                           quantiles=None,
                           hist_kwargs={"linewidth": 2.},      
                           contour_kwargs={"alpha": 1},
                           figsize=(6,4), **default_kwargs)

    corner.corner(data_hi_star, 
                            titles = title_labels, labels = labels,
                            color=ska_color_intermediate, alpha=0.75,
                            fill_contours=False,
                            show_titles=False,
                            title_quantiles=[0.16, 0.5, 0.84],
                            quantiles=[0.16, 0.84],
                            hist_kwargs={"linewidth": 2.},
                            range=[(40, 100), (0, 1), (1, 5), (2.5, 9), (0.75, 1.25), (0, 3.5), (-1, 1)],  # one tuple per parameter
                            rect=[0, 0, 1, 1], 
                            fig=figure, **default_kwargs)


    for ax in figure.get_axes():
        ax.tick_params(axis = 'both', labelsize = lbl_size)

    figure.subplots_adjust(wspace=0, hspace=0)
    legend_elements = [
        Line2D(
        [0], [0],
        color=ska_color_pink,
        lw=3.5,
        linestyle="solid",
        label=r"\rm Injected value"
    ),
    Patch(facecolor='none', edgecolor=ska_color_purple, linewidth=2.5, label=r"\rm Clustered GWs - {\textrm{H\textsc{i}}}", alpha=1),
    Patch(facecolor='none', edgecolor=ska_color_intermediate, linewidth=2.5, label=r"\rm Clustered GWs$^*$ - {\textrm{H\textsc{i}}}$^*$", alpha=0.75),
    ]

    ref_samples = np.asarray(ref_data)
    ndim = ref_samples.shape[1]
    axes = np.asarray(figure.axes).reshape(ndim, ndim)

    if default_kwargs.get("reverse", False):
        axes = axes[::-1, ::-1]

    for i in range(ndim):
        ax = axes[i, i]

        # Find the two histogram outlines.
        histograms = list(ax.patches)
        if not histograms:  # Handle smooth1d histograms.
            histograms = [
                line for line in ax.lines
                if np.ptp(line.get_xdata()) > 0
            ]

        histogram = histograms[0]  # ref_data is plotted second.

        lo, hi = corner.quantile(
            ref_samples[:, i],
            [0.16, 0.84],
            weights=default_kwargs.get("weights"),
        )

        if isinstance(histogram, Line2D):
            x, y = histogram.get_data()
            vertices = np.vstack([
                (x[0], 0),
                np.column_stack([x, y]),
                (x[-1], 0),
            ])
            clip = Polygon(vertices, closed=True, transform=ax.transData)
        else:
            clip = histogram

        shade = ax.axvspan(
            lo, hi,
            color=ska_color_purple,
            alpha=0.25,
            linewidth=0,
            zorder=0.5,
        )
        shade.set_clip_path(clip)


    figure.legend(handles=legend_elements, bbox_to_anchor=(0.95, 0.8), fontsize=21, frameon=False)
    plt.subplots_adjust(left=0.2, bottom=0.2, right=0.95, top=0.95)
    plt.savefig(f'corner_{label}_bias_full.pdf', bbox_inches='tight')
    plt.show()
```

```python
params = ['H0', 'Om0', 'gamma', 'kappa', 'zp', 'bGW', 'alphaGW']
injected_values = [67.7, 0.308, 2.7, 6, 1, 1, 0]
```

```python
ref_data = bilby.result.read_in_result('../../results/cosmoPEs_snr150_rs1_nside16_HIHI_bias_params/label_result.json')
hi_star_data = bilby.result.read_in_result('../../results/cosmoPEs_snr150_rs1_nside16_starHI_HI_varying_bias/label_result.json')
```

```python
fig = plot_corner(ref_data.posterior[params], hi_star_data.posterior[params], 'hi_star', injected_values)
```
