## N-body maps

```python
ska_color_pink = '#d60068'
ska_color_intermediate = '#840068'
ska_color_purple = '#180068'
radio_color = '#f5e15f'
```

```python
nside = 16
npix = hp.nside2npix(nside)
lmax = 3*nside 

# redhsift range for reference
zmin = 0.005
zmax = 3.025

H0_value = 67.7
# Reference cosmology
cosmo_ref = icarogw.cosmology.astropycosmology(zmax=20.)
cosmo_ref.build_cosmology(FlatLambdaCDM(H0=H0_value,Om0=0.308)) 
```

```python
PATH_TO_DENSITY_MATRIX = '../../hi_data/'
density_matrix_OV = pd.read_hdf(PATH_TO_DENSITY_MATRIX + 'map4icaro_lmax' + str(lmax) + '_zmin' + str(zmin) + '_zmax' + str(zmax) + '_nside' + str(nside) + '.hdf5').to_numpy()
density_matrix_HI = pd.read_hdf(PATH_TO_DENSITY_MATRIX + 'map4icaro_lmax' + str(lmax) + '_zmin' + str(zmin) + '_zmax' + str(zmax) + '_nside' + str(nside) + '_HI.hdf5').to_numpy()
density_matrix_star = pd.read_hdf(PATH_TO_DENSITY_MATRIX + 'map4icaro_lmax' + str(lmax) + '_zmin' + str(zmin) + '_zmax' + str(zmax) + '_nside' + str(nside) + '_star.hdf5').to_numpy()
```

```python
redshift_grid = np.array([0.005, 0.015, 0.035, 0.06 , 0.085, 0.11 , 0.135, 0.17 , 0.205,
        0.24 , 0.28 , 0.32 , 0.36 , 0.405, 0.455, 0.505, 0.56 , 0.62 ,
        0.68 , 0.745, 0.82 , 0.895, 0.975, 1.06 , 1.155, 1.26 , 1.365,
        1.485, 1.615, 1.755, 1.905, 2.05 , 2.205, 2.39 , 2.585, 2.795,
        3.025])
pixel_grid = np.arange(0, npix, 1).astype(int) # generates an array of (integer) numbers from 0 to npix-1
HI_OV = icarogw.HI.HI_map(redshift_grid,pixel_grid,density_matrix_OV, bGW=1, alphaGW=0)
HI_HI = icarogw.HI.HI_map(redshift_grid,pixel_grid,density_matrix_HI, bGW=1, alphaGW=0)
HI_star = icarogw.HI.HI_map(redshift_grid,pixel_grid,density_matrix_star, bGW=1, alphaGW=0)
```

```python
from scipy.ndimage import gaussian_filter1d
gaussian_sigma = 6.5
```

```python
fig = plt.figure(figsize=(6,4))
my_linewidth = 2.5
#pixx1 = 1203 # pick a pixel number (in 0 to npix-1) to plot the density of hydrogen as function of redshift#
pixx1 = 23
pixx2 = 1203
z_plot = np.linspace(0.005,10,1000)
pixels_array = pixx1*np.ones_like(z_plot)
pixels_array2 = pixx2*np.ones_like(z_plot)
gc_OV = HI_OV.drho_dzdomega(z_plot,pixels_array,cosmo_ref)
gc_HI = HI_HI.drho_dzdomega(z_plot,pixels_array,cosmo_ref)
gc_star = HI_star.drho_dzdomega(z_plot,pixels_array,cosmo_ref)
gc_OV2 = HI_OV.drho_dzdomega(z_plot,pixels_array2,cosmo_ref)
gc_HI2 = HI_HI.drho_dzdomega(z_plot,pixels_array2,cosmo_ref)
# smooth the data points
gc_smooth = gaussian_filter1d(gc_OV, sigma=gaussian_sigma)
gc_smooth_HI = gaussian_filter1d(gc_HI, sigma=gaussian_sigma)
gc_smooth_star = gaussian_filter1d(gc_star, sigma=gaussian_sigma)
#plt.semilogx(z_plot,gc_smooth,label=r'\rm RA = {:.0f}$^{{\circ}}$, Dec = {:.0f}$^{{\circ}}$'.format(ra, dec), linewidth=my_linewidth, color='black', linestyle='dotted')
plt.semilogx(z_plot,gc_OV, linewidth=my_linewidth, color=ska_color_pink, linestyle='dotted')
plt.semilogx(z_plot,gc_HI, linewidth=my_linewidth, color=ska_color_pink, linestyle='solid', alpha=0.75)
plt.semilogx(z_plot,gc_OV2, linewidth=my_linewidth, color=ska_color_purple, linestyle='dotted', alpha=0.75)
plt.semilogx(z_plot,gc_HI2, linewidth=my_linewidth, color=ska_color_purple, linestyle='solid', alpha=0.75)

#plt.semilogx(z_plot,gc_smooth_star, linewidth=my_linewidth, color=ska_color_purple, linestyle='solid', alpha=0.75)
plt.xscale('log')
#plt.yscale('log')
plt.xlim(0.005, 10)
#gc = HI.drho_dzdomega(z_plot,pixels_array,cosmo_ref,average=True)
#plt.plot(z_plot,gc,label = 'Averaged over sky', linestyle='dashed', linewidth=2.5)
#plt.yscale('log')
plt.xlabel(r'\rm Redshift $z$', fontsize =16)
plt.ylabel(r'$\rho_{\rm BH} / \bar{\rho}_{\rm BH}$', fontsize =16)
#plt.text(0.4, 0.855, r'\rm RA = {:.0f}$^{{\circ}}$, Dec = {:.0f}$^{{\circ}}$'.format(ra, dec), fontsize=14)

ax = plt.gca()

# Preserve the existing RA/Dec legend
position_legend = ax.legend(fontsize=14, frameon=False)
ax.add_artist(position_legend)

# Add a separate legend above the axes
style_handles = [
    Line2D([0], [0], color=color, lw=my_linewidth,
           linestyle=style, label=label, alpha=alpha)
    for style, color, alpha, label in [
        ('dotted', 'gray', 1., r'\rm {\textrm{H\textsc{i}}}'),
        ('solid', 'gray', 0.75, r'\rm {\textrm{H\textsc{i}}}$^*$'),
        #('solid', ska_color_purple, 0.75, r'\rm star'),
    ]
]

ax.legend(
    handles=style_handles,
    loc='lower center',
    bbox_to_anchor=(0.5, 0.98),
    ncol=1,
    fontsize=16,
    frameon=False,
)

plt.savefig('HI_density_profile_smooth_plus_bias.png', dpi=300, bbox_inches='tight')
plt.show()

```
